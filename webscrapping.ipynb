```python
import requests
from bs4 import BeautifulSoup
import random
from IPython.display import display, Image

def get_news_channel():
    """Function to get a valid news channel from the user."""
    channel_name = input("Enter the news channel name [bbc, the hindu, the deccan chronicle, cnn,e tv,tv9,abn]: ").lower()
    channel_map = {
        "bbc": ('https://www.bbc.com/news', 'h2'),
        "the hindu": ('https://www.thehindu.com/', 'h3'),
        "the deccan chronicle": ('https://www.deccanchronicle.com/', 'h2'),
        "cnn": ('https://edition.cnn.com/', 'h2'),
        "e tv":('https://www.eenadu.net/andhra-pradesh','h2'),
        "tv9":('https://tv9telugu.com/andhra-pradesh','h3'),
        "abn":('https://www.andhrajyothy.com/latest-news/page/3','h3')
    }

    if channel_name in channel_map:
        return channel_map[channel_name]
    else:
        print("Invalid news channel name.")
        return None, None

def fetch_headlines(URL, headline_tag, num_headlines):
    """Function to fetch headlines from the given URL."""
    try:
        print("Fetching news... Please wait.")
        response = requests.get(URL)
        response.raise_for_status()  # Raise an error for bad responses
    except requests.RequestException as e:
        print(f"An error occurred: {e}")
        return []

    soup = BeautifulSoup(response.content, 'html.parser')
    return soup.find_all(headline_tag)[:num_headlines]  # Limit to user-specified number

def display_headlines(headlines, URL):
    """Function to display fetched headlines with images."""
    colors = [
        "\033[91m", "\033[92m", "\033[93m", "\033[94m",
    ]
    reset_color = "\033[0m"
    bold_format = "\033[1m"

    for h in headlines:
        color = random.choice(colors)
        headline_text = h.text.strip().upper()
        print(f"{color}{bold_format}HEADLINE: {headline_text}{reset_color}")

        # Attempt to find an image
        img_tag = h.find_previous('img')
        if img_tag and 'src' in img_tag.attrs:
            img_src = img_tag['src']
            if img_src.startswith('/'):
                img_src = URL + img_src
            print(img_src)
            display(Image(url=img_src))
        else:
            print("No image found for this headline.")

        print("\n" * 2)

def main():
    """Main function to run the news scraper."""
    while True:
        URL, headline_tag = get_news_channel()
        if URL is None:
            continue  # Prompt again if the channel is invalid

        num_headlines = input("How many headlines would you like to see? (Enter a number): ")
        try:
            num_headlines = int(num_headlines)
        except ValueError:
            print("Please enter a valid number.")
            continue

        headlines = fetch_headlines(URL, headline_tag, num_headlines)

        if headlines:
            display_headlines(headlines, URL)
        else:
            print("No headlines found.")

        # Ask if the user wants to scrape again
        repeat = input("Would you like to scrape another channel? (yes/no): ").lower()
        if repeat != 'yes':
            print("Exiting the news scraper. Goodbye!")
            break

if __name__ == "__main__":
    main()
```
